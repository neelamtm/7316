**10. Plot the image.**
This is a tricky one because the image (as you see above) is huge. In fact, it is too big to fit in Google Colab's memory. So...read a `rasterio.windows.Window` of the image. See the documentation. I suggest a window not larger than 3000 x 2000. BUT, you can use AI for this question. You can use `matplotlib` or `ee` and `geemap`.

````python
from rasterio.windows import Window
import matplotlib.pyplot as plt
import numpy as np
````

````python
import ee
import geemap

ee.Authenticate()
ee.Initialize(project='project-7316')
Map = geemap.Map()
Map
````

````python
# Define window size (3000*2000)
window_width = 3000
window_height = 2000
````

````python
# Calculate offsets to get a central window
col_off = (rast.width - window_width) // 2
row_off = (rast.height - window_height) // 2

# Create window object
read_window = Window(col_off, row_off, window_width, window_height)

# Read data from new defined window
img_window = rast.read(window=read_window)
print(img_window.shape)
````

````python
# Transpose the array for matplotlib (from bands, height, width to height, width, bands)
# And select the first 3 bands for RGB display
# If it's a grayscale image, img_window[0] could be directly plotted.
if img_window.shape[0] >= 3:
    img_display = np.transpose(img_window[:3, :, :], (1, 2, 0))
else:
    img_display = img_window[0, :, :]

# Plot the image window
plt.figure(figsize=(10, 10))
plt.imshow(img_display)
plt.title(f'Image Window (offset_col={col_off}, offset_row={row_off}, width={window_width}, height={window_height})')
plt.axis('off')
plt.show()
````
